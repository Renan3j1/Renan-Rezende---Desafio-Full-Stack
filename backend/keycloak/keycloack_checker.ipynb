```python
import requests

response = requests.post(
    "http://localhost:8080/realms/notescreen/protocol/openid-connect/token",
    data={
        "client_id": "notescreen-test",
        "grant_type": "password",
        "username": "tester",
        "password": "teste123",
    },
)

response.raise_for_status()
token = response.json()["access_token"]
print("Login OK " + token)
```
