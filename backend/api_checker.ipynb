```python
import requests
```

```python
response = requests.get(
    "http://localhost:8000/api/v1/notes?site=Braga, Braga e Melo&page=1&page_size=10"
)
response.raise_for_status()
data = response.json()
data
```

```python
payload = {
    "site": "Braga, Braga e Melo",
    "equipment": "Equipamento 1",
    "variable": "Grade",
    "timestamp": "2024-06-01",
    "author": "Renan Rezende",
    "message": "Esta é uma nota de teste.",
}

response = requests.post(
    "http://localhost:8000/api/v1/notes",
    json=payload,
)
response.raise_for_status()

created_note = response.json()
created_note
```

```python
response = requests.delete(
    "http://localhost:8000/api/v1/notes/6492c8f7-2a93-455f-9117-8f4aff988ec8",
    json=payload,
)
response.raise_for_status()

deleted_note = response.json()
deleted_note
```

```python
payload = {
    "site": "Rezendech",
    "equipment": "Savignado",
    "variable": "Cerca",
    "timestamp": "2024-06-06",
    "author": "Renan Rezende",
    "message": "Esta é uma nota de alterada.",
}

response = requests.put(
    "http://localhost:8000/api/v1/notes/2b84b636-9b78-48c1-a67b-431ff0af8e91",
    json=payload,
)
response.raise_for_status()

updated_note = response.json()
updated_note
```
